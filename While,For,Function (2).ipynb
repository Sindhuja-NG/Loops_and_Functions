```python
# Number Guessing Game

import random

secret_number = random.randint(1, 10)
attempt=3

while attempt>0:
 userguess=int(input("Enter the number:"))
 if(userguess>10):
  print("Invalid input range,Give input range between 1 to 10")
  continue
 elif(userguess>secret_number):
   print("your guess is too high,try again")
 elif(userguess<secret_number):
   print("your guess is too low,try again")
 elif(userguess==secret_number):
   print("wowiee!!!your guess is correct")
   break
   attempt -=1
else:
  print("Better luck next time")



```

```python
#Create a Python program that generates and prints a multiplication table (from 1 to 10) for a given number using a for loop and the range function.

num=int(input("Enter the number:"))
for i in range(1,11):
  print(num,"x",i,"=",num*i)
```

```python
#Function

def calculate_bmi(weight, height):
  bmi = weight / (height ** 2)
  return bmi

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = calculate_bmi(weight, height)
print("Your BMI is:", round(bmi, 1))
```
